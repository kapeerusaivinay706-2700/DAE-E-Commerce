```python
# ============================================================
# E-COMMERCE DATA ANALYSIS - GOOGLE COLAB ONLY
# ============================================================

!pip -q install openpyxl plotly

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
from IPython.display import display, HTML
import plotly.express as px

# ============================================================
# 1. UPLOAD DATASET
# ============================================================

print("📂 Upload your E-Commerce CSV or Excel dataset")
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

if file_name.lower().endswith(".csv"):
    df = pd.read_csv(file_name)
elif file_name.lower().endswith((".xlsx", ".xls")):
    df = pd.read_excel(file_name)
else:
    raise ValueError("Please upload a CSV or Excel file.")

print("\n✅ Dataset loaded successfully!")
print("File:", file_name)

# ============================================================
# 2. STANDARDIZE COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(" ", "_", regex=False)
)

# Automatically recognize common column names
rename_map = {}

for col in df.columns:
    c = col.lower().replace("_", "").replace(" ", "")

    if c in ["orderid"]:
        rename_map[col] = "Order_ID"

    elif c in ["orderdate", "date"]:
        rename_map[col] = "Order_Date"

    elif c in ["customerid"]:
        rename_map[col] = "Customer_ID"

    elif c in ["customername", "name"]:
        rename_map[col] = "Customer_Name"

    elif c in ["product", "productname"]:
        rename_map[col] = "Product_Name"

    elif c in ["category"]:
        rename_map[col] = "Category"

    elif c in ["quantity", "qty"]:
        rename_map[col] = "Quantity"

    elif c in ["price", "unitprice"]:
        rename_map[col] = "Price"

    elif c in ["paymentmethod", "payment"]:
        rename_map[col] = "Payment_Method"

    elif c in ["orderstatus", "status"]:
        rename_map[col] = "Order_Status"

df.rename(columns=rename_map, inplace=True)

print("\n📋 Columns detected:")
print(list(df.columns))

# ============================================================
# 3. DATASET OUTPUT
# ============================================================

print("\n" + "="*70)
print("1. DATASET OUTPUT")
print("="*70)

# Raw dataset
print("\n1.1 RAW DATASET")
display(df)

# df.head()
print("\n1.2 df.head()")
display(df.head())

# Dataset structure
print("\n1.3 DATASET STRUCTURE")

structure = pd.DataFrame({
    "Column": df.columns,
    "Non-Null Count": [
        df[col].notna().sum()
        for col in df.columns
    ],
    "Data Type": [
        str(df[col].dtype)
        for col in df.columns
    ]
})

display(structure)

print("\nDataset Shape:", df.shape)

# ============================================================
# 4. PREPROCESSING
# ============================================================

print("\n" + "="*70)
print("2. PREPROCESSING OUTPUT")
print("="*70)

raw_df = df.copy()

# Missing values
print("\n2.1 MISSING-VALUE RESULTS")

missing_values = pd.DataFrame({
    "Column": raw_df.columns,
    "Missing Values": [
        raw_df[col].isna().sum()
        for col in raw_df.columns
    ]
})

display(missing_values)

# Duplicate count
duplicate_count = raw_df.duplicated().sum()

print("\n2.2 DUPLICATE COUNT")
print("Duplicate records:", duplicate_count)

# Convert date
if "Order_Date" in df.columns:
    df["Order_Date"] = pd.to_datetime(
        df["Order_Date"],
        errors="coerce"
    )

# Convert numerical columns
if "Quantity" in df.columns:
    df["Quantity"] = pd.to_numeric(
        df["Quantity"],
        errors="coerce"
    )

if "Price" in df.columns:
    df["Price"] = pd.to_numeric(
        df["Price"],
        errors="coerce"
    )

# Remove duplicates
cleaned_df = df.drop_duplicates().copy()

# Remove duplicate Order IDs
if "Order_ID" in cleaned_df.columns:
    cleaned_df = cleaned_df.drop_duplicates(
        subset=["Order_ID"]
    )

# Fill Quantity
if "Quantity" in cleaned_df.columns:
    cleaned_df["Quantity"] = cleaned_df["Quantity"].fillna(
        cleaned_df["Quantity"].median()
    )

# Fill Category
if "Category" in cleaned_df.columns:
    cleaned_df["Category"] = cleaned_df["Category"].fillna(
        "Unknown"
    )

# Fill Payment Method
if "Payment_Method" in cleaned_df.columns:
    if cleaned_df["Payment_Method"].notna().any():
        mode_value = cleaned_df["Payment_Method"].mode().iloc[0]
        cleaned_df["Payment_Method"] = (
            cleaned_df["Payment_Method"].fillna(mode_value)
        )

# Remove rows missing important information
required_for_cleaning = []

for col in ["Order_ID", "Order_Date", "Price"]:
    if col in cleaned_df.columns:
        required_for_cleaning.append(col)

if required_for_cleaning:
    cleaned_df = cleaned_df.dropna(
        subset=required_for_cleaning
    )

# Remove invalid values
if "Quantity" in cleaned_df.columns:
    cleaned_df = cleaned_df[
        cleaned_df["Quantity"] > 0
    ]

if "Price" in cleaned_df.columns:
    cleaned_df = cleaned_df[
        cleaned_df["Price"] > 0
    ]

# Calculate Sales
if "Quantity" in cleaned_df.columns and "Price" in cleaned_df.columns:
    cleaned_df["Sales"] = (
        cleaned_df["Quantity"] *
        cleaned_df["Price"]
    )

original_count = len(raw_df)
cleaned_count = len(cleaned_df)
removed_count = original_count - cleaned_count

# Cleaned dataset
print("\n2.3 CLEANED DATASET")
display(cleaned_df)

# Record count
print("\n2.4 RECORD COUNT AFTER CLEANING")

print("Original Records :", original_count)
print("Cleaned Records  :", cleaned_count)
print("Records Removed  :", removed_count)

# ============================================================
# 5. VISUALIZATION
# ============================================================

print("\n" + "="*70)
print("3. VISUALIZATION OUTPUT")
print("="*70)

# Plot style
sns.set_theme(style="whitegrid")

# ------------------------------------------------------------
# 3.1 CATEGORY-WISE SALES
# ------------------------------------------------------------

if "Category" in cleaned_df.columns:

    category_sales = (
        cleaned_df
        .groupby("Category")["Sales"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\n3.1 CATEGORY-WISE SALES CHART")

    plt.figure(figsize=(10,5))
    category_sales.plot(kind="bar")

    plt.title("Category-wise Sales")
    plt.xlabel("Category")
    plt.ylabel("Sales")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# 3.2 PAYMENT METHOD PIE CHART
# ------------------------------------------------------------

if "Payment_Method" in cleaned_df.columns:

    payment_counts = (
        cleaned_df["Payment_Method"]
        .value_counts()
    )

    print("\n3.2 PAYMENT-METHOD PIE CHART")

    plt.figure(figsize=(7,7))

    plt.pie(
        payment_counts.values,
        labels=payment_counts.index,
        autopct="%1.1f%%",
        startangle=90
    )

    plt.title("Payment Method Distribution")
    plt.show()

# ------------------------------------------------------------
# 3.3 MONTHLY SALES
# ------------------------------------------------------------

if "Order_Date" in cleaned_df.columns:

    cleaned_df["Month"] = (
        cleaned_df["Order_Date"]
        .dt.to_period("M")
        .astype(str)
    )

    monthly_sales = (
        cleaned_df
        .groupby("Month")["Sales"]
        .sum()
    )

    print("\n3.3 MONTHLY SALES LINE CHART")

    plt.figure(figsize=(12,5))

    plt.plot(
        monthly_sales.index,
        monthly_sales.values,
        marker="o"
    )

    plt.title("Monthly Sales Trend")
    plt.xlabel("Month")
    plt.ylabel("Sales")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()

# ------------------------------------------------------------
# 3.4 ORDER-VALUE HISTOGRAM
# ------------------------------------------------------------

print("\n3.4 ORDER-VALUE HISTOGRAM")

plt.figure(figsize=(10,5))

plt.hist(
    cleaned_df["Sales"],
    bins=30
)

plt.title("Order Value Distribution")
plt.xlabel("Order Value / Sales")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 3.5 SALES BOX PLOT
# ------------------------------------------------------------

print("\n3.5 SALES BOX PLOT")

plt.figure(figsize=(8,5))

plt.boxplot(
    cleaned_df["Sales"],
    vert=True
)

plt.title("Sales Box Plot")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 3.6 QUANTITY VS SALES
# ------------------------------------------------------------

print("\n3.6 QUANTITY VS SALES SCATTER PLOT")

plt.figure(figsize=(10,6))

if "Category" in cleaned_df.columns:

    for category in cleaned_df["Category"].unique():

        temp = cleaned_df[
            cleaned_df["Category"] == category
        ]

        plt.scatter(
            temp["Quantity"],
            temp["Sales"],
            label=category,
            alpha=0.6
        )

    plt.legend()

else:

    plt.scatter(
        cleaned_df["Quantity"],
        cleaned_df["Sales"],
        alpha=0.6
    )

plt.title("Quantity vs Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 3.7 TOP PRODUCTS
# ------------------------------------------------------------

if "Product_Name" in cleaned_df.columns:

    top_products = (
        cleaned_df
        .groupby("Product_Name")["Sales"]
        .sum()
        .sort_values(ascending=False)
        .head(10)
        .sort_values()
    )

    print("\n3.7 TOP-PRODUCTS CHART")

    plt.figure(figsize=(10,6))

    top_products.plot(
        kind="barh"
    )

    plt.title("Top 10 Products by Sales")
    plt.xlabel("Sales")
    plt.ylabel("Product")

    plt.tight_layout()
    plt.show()

# ============================================================
# 6. ANALYSIS OUTPUT
# ============================================================

print("\n" + "="*70)
print("4. ANALYSIS OUTPUT")
print("="*70)

# ------------------------------------------------------------
# TOTAL SALES
# ------------------------------------------------------------

total_sales = cleaned_df["Sales"].sum()

# ------------------------------------------------------------
# TOTAL ORDERS
# ------------------------------------------------------------

if "Order_ID" in cleaned_df.columns:

    total_orders = cleaned_df["Order_ID"].nunique()

else:

    total_orders = len(cleaned_df)

# ------------------------------------------------------------
# AVERAGE ORDER VALUE
# ------------------------------------------------------------

average_order_value = (
    total_sales / total_orders
    if total_orders > 0
    else 0
)

print("\n4.1 TOTAL SALES")
print(f"₹ {total_sales:,.2f}")

print("\n4.2 TOTAL ORDERS")
print(f"{total_orders:,}")

print("\n4.3 AVERAGE ORDER VALUE")
print(f"₹ {average_order_value:,.2f}")

# ------------------------------------------------------------
# CATEGORY SUMMARY
# ------------------------------------------------------------

if "Category" in cleaned_df.columns:

    print("\n4.4 CATEGORY SUMMARY")

    category_summary = (
        cleaned_df
        .groupby("Category")
        .agg(
            Total_Sales=("Sales", "sum"),
            Total_Orders=("Order_ID", "nunique"),
            Quantity_Sold=("Quantity", "sum")
        )
        .reset_index()
        .sort_values(
            "Total_Sales",
            ascending=False
        )
    )

    display(category_summary)

# ------------------------------------------------------------
# TOP PRODUCTS
# ------------------------------------------------------------

if "Product_Name" in cleaned_df.columns:

    print("\n4.5 TOP PRODUCTS")

    top_products_table = (
        cleaned_df
        .groupby("Product_Name")
        .agg(
            Total_Sales=("Sales", "sum"),
            Quantity_Sold=("Quantity", "sum"),
            Orders=("Order_ID", "nunique")
        )
        .reset_index()
        .sort_values(
            "Total_Sales",
            ascending=False
        )
        .head(10)
    )

    display(top_products_table)

# ------------------------------------------------------------
# TOP CUSTOMERS
# ------------------------------------------------------------

if "Customer_ID" in cleaned_df.columns:

    print("\n4.6 TOP CUSTOMERS")

    customer_group = [
        "Customer_ID"
    ]

    if "Customer_Name" in cleaned_df.columns:
        customer_group.append("Customer_Name")

    top_customers = (
        cleaned_df
        .groupby(customer_group)
        .agg(
            Total_Purchase=("Sales", "sum"),
            Total_Orders=("Order_ID", "nunique")
        )
        .reset_index()
        .sort_values(
            "Total_Purchase",
            ascending=False
        )
        .head(10)
    )

    display(top_customers)

# ------------------------------------------------------------
# ORDER STATUS
# ------------------------------------------------------------

if "Order_Status" in cleaned_df.columns:

    print("\n4.7 ORDER-STATUS ANALYSIS")

    status_analysis = (
        cleaned_df
        .groupby("Order_Status")
        .size()
        .reset_index(name="Orders")
    )

    status_analysis["Percentage"] = (
        status_analysis["Orders"]
        / status_analysis["Orders"].sum()
        * 100
    ).round(2)

    display(status_analysis)

# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "="*70)
print("FINAL PROJECT SUMMARY")
print("="*70)

print(f"""
Total Sales          : ₹ {total_sales:,.2f}
Total Orders         : {total_orders:,}
Average Order Value  : ₹ {average_order_value:,.2f}
Original Records     : {original_count:,}
Cleaned Records      : {cleaned_count:,}
Duplicate Records    : {duplicate_count:,}
Records Removed      : {removed_count:,}
""")

print("✅ E-Commerce Data Analysis completed successfully!")
```
